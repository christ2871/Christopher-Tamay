## Section 2. Three words the rest of the course depends on

People use threat, vulnerability and risk as if they were the same word. They are not, and
every later lab in this course assumes you can tell them apart.

- A **threat** is an actor or an event with the potential to cause harm. It exists whether or
  not you are exposed to it.
- A **vulnerability** is a weakness a threat could use. It exists whether or not anyone is
  currently trying to use it.
- A **risk** is what the two produce together, expressed as a likelihood and an impact. A risk
  is always a possible loss, never a thing.

Below are eight observations from one small campus scenario: the library self-service print
station and the systems behind it. Label each one `threat`, `vulnerability` or `risk`.

Run the next cell to read them.
